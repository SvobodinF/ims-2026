## Подготовка к заданиям

Создание исходного DataFrame для выполнения заданий.

```python
import pandas as pd
import numpy as np

data = {
    'ID': range(1, 11),
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva',
             'Frank', 'Grace', 'Helen', 'Ivan', 'Julia'],
    'Age': [23, 35, 29, 40, 28, 45, 32, 27, 38, 31],
    'City': ['Moscow', 'SPb', 'Moscow', 'Kazan', 'SPb',
             'Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb'],
    'Salary': [50000, 70000, 60000, 80000, 55000,
               90000, 65000, 58000, 75000, 62000]
}

df = pd.DataFrame(data)
print(df)
```

### Задание 1. Вывести названия всех столбцов (`.columns`)

```python
print(df.columns)
```

### Задание 2. Выбрать все строки, кроме последних двух

```python
print(df.iloc[:-2])
```

### Задание 3. Выбрать строки, где `Age` меньше среднего возраста

```python
mean_age = df['Age'].mean()
print('Средний возраст:', mean_age)
print(df[df['Age'] < mean_age])
```

### Задание 4. Найти количество уникальных городов

```python
print(df['City'].nunique())
```

### Задание 5. Создать столбец `Rank` — ранг по `Salary` (1 — наибольшая зарплата), используя `.rank(ascending=False)`

```python
df['Rank'] = df['Salary'].rank(ascending=False)
print(df[['Name', 'Salary', 'Rank']])
```
