### 3.2 Выбор строк: `.loc[]` и `.iloc[]`

#### `.loc[]` — выбор по меткам (label-based)
Использует имена строк и столбцов. Срезы с `.loc[]` включают обе границы (начало и конец).

```python
# Выбор строки с индексом 0
print(df.loc[0])

# Выбор конкретной ячейки
print(df.loc[0, 'Name'])

# Выбор диапазона строк (включительно!)
print(df.loc[0:2, ['Name', 'Age']])
```

#### `.iloc[]` — выбор по позициям (integer-based)
Использует целочисленные индексы (как в списках Python). Срезы с `.iloc[]` работают как в Python — конец не включается.

```python
# Первая строка (индекс 0)
print(df.iloc[0])

# Первая строка, второй столбец
print(df.iloc[0, 1])

# Первые две строки, первые два столбца
print(df.iloc[0:2, 0:2])

# Все строки, столбцы с 1 по 2 (не включая 2)
print(df.iloc[:, 1:3])
```

### 3.3 Условная выборка (фильтрация)

```python
# Выбрать строки, где Age > 40
print(df[df['Age'] > 40])

# Несколько условий (используйте & для "И", | для "ИЛИ")
print(df[(df['Age'] > 30) & (df['Pclass'] == 1)])

# Выбор по значениям из списка
print(df[df['Embarked'].isin(['C', 'Q'])])
```

### 3.4 Присваивание значений

```python
# Создать новый столбец
df['FareRounded'] = df['Fare'].round()

# Изменить значения по условию
df.loc[df['Age'] < 18, 'AgeGroup'] = 'Ребенок'
df.loc[df['Age'] >= 18, 'AgeGroup'] = 'Взрослый'

print(df[['Name', 'Fare', 'FareRounded', 'Age', 'AgeGroup']].head())
```

## 4. Функции сводки и отображения (Summary Functions & Maps)

### 4.1 Статистические функции

```python
# Описательная статистика для числовых столбцов
print(df.describe())

# Среднее значение и медиана
print(df['Age'].mean())
print(df['Age'].median())

# Минимум и максимум
print(df['Fare'].min())
print(df['Fare'].max())

# Сумма и стандартное отклонение
print(df['Fare'].sum())
print(df['Age'].std())

# Количество уникальных значений и список уникальных значений
print(df['Embarked'].nunique())
print(df['Embarked'].unique())

# Подсчет частоты значений
print(df['Pclass'].value_counts())
```

### 4.2 Применение функций: `.map()`, `.apply()`

`.map()` — для Series (применение функции к каждому элементу)

```python
# Перевод пола на русский
df['Sex_RU'] = df['Sex'].map({'male': 'Мужчина', 'female': 'Женщина'})

# Категоризация по возрасту
def categorize_age(age):
    if pd.isna(age):
        return 'Неизвестно'
    elif age < 18:
        return 'Ребенок'
    elif age < 60:
        return 'Взрослый'
    else:
        return 'Пожилой'

df['AgeCategory'] = df['Age'].map(categorize_age)

print(df[['Name', 'Sex', 'Sex_RU', 'Age', 'AgeCategory']].head())
```

`.apply()` — для DataFrame (применение функции к строкам/столбцам)

```python
# Применение к столбцу (аналог map)
df['FareDoubled'] = df['Fare'].apply(lambda x: x * 2)

# Сложная функция, применяемая построчно
def calculate_score(row):
    age = row['Age'] if pd.notna(row['Age']) else 0
    return row['Fare'] * age / 100

df['Score'] = df.apply(calculate_score, axis=1)

print(df[['Name', 'Fare', 'FareDoubled', 'Age', 'Score']].head())
```
